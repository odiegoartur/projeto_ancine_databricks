```python
## criando o df_bronze usando a tabela da camada bronze 

df_bronze = spark.read.table("projeto_ancine.bronze.bilheteria_distribuidoras")

display(df_bronze.limit(5))
```

```python
## excluindo colunas desnecessárias 

df_transformado = df_bronze
display(df_transformado.limit(5))

df_transformado = df_transformado.drop("NR_PROTOCOLO_ENVIO","DATA_HORA_ENVIO_PROTOCOLO")
df_transformado.printSchema()
```

```python
## corrigindo o nome das columas para minusculo

## executa um for nas colunas de df_transformado. aplica o .lower nas colunas uma por vez chamando elas de nome. salva isso na lista colunas_minusc
colunas_minusc = [nome.lower() for nome in df_transformado.columns]

## substitui os nomes das colunas pelos nomes que estão em colunas_minusc
df_transformado = df_transformado.toDF(*colunas_minusc)

## verificando o resultado
df_transformado.printSchema()
```

```python
## conferindo linhas nulas

df_linhas_invalidas = df_transformado.filter(col("titulo_original").isNull() | (col("titulo_original") == ""))

total_invalidas = df_linhas_invalidas.count()
print(f"Total de linhas sem título original: {total_invalidas}")

display(df_linhas_invalidas)
```
